```python
class MultiplicationTable:
    """Class to print the multiplication table of a given number.
    Attributes:
        number (int): The number for which the table is generated.
    """
    def __init__(self, number):
        self.number = number

    def print_table(self):
        """Prints the multiplication table from 1 to 10."""
        for i in range(1, 11):
            print(f"{self.number} * {i} = {self.number * i}")
table = MultiplicationTable(9)
table.print_table()
```

```python
class TwinPrimesFinder:
    """Class to find and print twin prime numbers below a given limit.
    Attributes:
        limit (int): The upper limit to find twin primes.
    """
    def __init__(self, limit=1000):
        self.limit = limit
    def is_prime(self, n):
        """Helper method to check if a number is prime."""
        if n < 2:
            return False
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False
        return True
    def print_twin_primes(self):
        """Prints all twin prime pairs less than the limit."""
        for first_num in range(2, self.limit - 2):
            second_num = first_num + 2
            if self.is_prime(first_num) and self.is_prime(second_num):
                print(f"Twin Prime: {first_num} and {second_num}")
finder = TwinPrimesFinder(1000)
finder.print_twin_primes()
```

```python
class PrimeFactorsCalculator:
    """Class to find the prime factors of a number.
    Attributes:
        number (int): The target number to factorize.
    """
    def __init__(self, number):
        self.number = number
    def get_prime_factors(self):
        """Calculates and returns a list of prime factors.
        Returns:
            list: List of prime factors.
        """
        factors = []
        n = self.number
        while n % 2 == 0:
            factors.append(2)
            n //= 2
        i = 3
        while i * i <= n:
            while n % i == 0:
                factors.append(i)
                n //= i
            i += 2
        if n > 2:
            factors.append(n)
        return factors
calculator = PrimeFactorsCalculator(56)
print(calculator.get_prime_factors())
```

```python
class DecimalToBinaryConverter:
    """Class to convert a decimal number to binary.
    Attributes:
        decimal_num (int): The decimal number to convert.
    """
    def __init__(self, decimal_num):
        self.decimal_num = decimal_num
    def convert(self):
        """Converts decimal number into binary string representation.
        Returns:
            str: Binary representation of the number.
        """
        if self.decimal_num == 0:
            return "0"
        return bin(self.decimal_num)[2:]
converter = DecimalToBinaryConverter(11)
print(converter.convert())
```

```python
class PerfectNumbersFinder:
    """Class to find all perfect numbers in a given range.
    Attributes:
        start (int): Lower bound of the range.
        end (int): Upper bound of the range.
    """
    def __init__(self, start, end):
        self.start = start
        self.end = end
    def is_perfect(self, num):
        """Checks if a given number is a perfect number."""
        if num <= 1:
            return False
        divisors_sum = 0
        for i in range(1, num):
            if num % i == 0:
                divisors_sum += i
        return divisors_sum == num
    def print_perfect_numbers(self):
        """Prints all perfect numbers found within the given range."""
        for num in range(self.start, self.end):
            if self.is_perfect(num):
                print(num)
finder = PerfectNumbersFinder(1, 100)
finder.print_perfect_numbers()
```
