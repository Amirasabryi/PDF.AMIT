```python
def train_model(model_name, *datasets, **hyperparameters):
    print(f"Training model: {model_name}")
    print("\n Datasets")
    for dataset in datasets:
        print("_", dataset)
    print("\n Hyperparameters")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")
```

```python
train_model(
    "fraude detection",
    "train.csv",
    "valid.csv",
    "test.csv",
    learning_rate = 0.01,
    batch_size = 32,
    epochs = 10,
    optimizer = "Adam",
    earlystopping = True
)
```
