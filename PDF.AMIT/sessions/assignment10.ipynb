```python
import numpy as np

# data
x = np.array([1, 2, 3, 4])
y = np.array([2, 2.8, 3.6, 4.5])

w = 0
b = 0
lr = 0.01

n = len(x)

for i in range(2):
    y_hat = w * x + b
    
    dw = (2/n) * sum((y_hat - y) * x)
    db = (2/n) * sum(y_hat - y)
    
    w = w - lr * dw
    b = b - lr * db
    
    print("iter:", i+1)
    print("w =", w)
    print("b =", b)
    print("dw =", dw)
    print("db =", db)
    print("---")
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y, color='red', label='Data Points')
plt.plot(x, w * x + b, color='blue', label='Prediction')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```
