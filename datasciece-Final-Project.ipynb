

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with ......median\mean..... for numeric and .other\mode.....for categorical

```python
# trying to fill missing values in my datset using for loop 

for col in gbf.columns:
    if gbf[col].dtype in ['float64', 'int64']:
        gbf[col].fillna(gbf[col].median(), inplace=True)
    else:
        gbf[col].fillna(gbf[col].mode()[0], inplace=True)
```

```python
#check the Nulls 
gbf.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more 

# print(gbf.member_gender.head(15))
gbf['member_gender'].head(15)
```

```python
# fill  dataset with missing values.
    
gbf['member_gender'].fillna('Other', inplace=True)
```

```python
# Make visualization to easy get insight
plt.figure(figsize=(10, 6), dpi=100)
numeric_cols = gbf.select_dtypes(include=['float64', 'int64']).columns
sns.heatmap(
    gbf[numeric_cols].corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap of Dataset Variables')
plt.show()
```
